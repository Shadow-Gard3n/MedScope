```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import json
from collections import defaultdict
import time
```

```python
df = pd.read_csv('medicine_dataset.csv', low_memory = False)
df
```

```python
df.info()
```

```python
df = df.fillna('')
df.info()
```

```python
DRUG_NAME_COL = 'name'
USAGE_COLS = [f'use{i}' for i in range(5)]
SIDEEFFECT_COLS = [f'sideEffect{i}' for i in range(42)]
```

```python
alternatives_lookup = defaultdict(list)
```

```python
start_time = time.time()
for _, row in df.iterrows():
    drug_name = str(row[DRUG_NAME_COL]).strip()
    if not drug_name:
        continue 
    # Get all unique side effects for this drug
    # We use a 'set' to automatically handle duplicates
    drug_side_effects = set()
    for col in SIDEEFFECT_COLS:
        effect = str(row[col]).strip().lower()
        if effect: # Only add it if it's not an empty string
            drug_side_effects.add(effect)
    
    # Create the info for this drug
    drug_info = {
        "name": drug_name,
        "effects": list(drug_side_effects) # Convert the set to a list
    }

    # Get all unique uses for this drug
    drug_usages = set()
    for col in USAGE_COLS:
        usage = str(row[col]).strip().lower()
        if usage: # Only add it if it's not an empty string
            drug_usages.add(usage)
    # For every use, add this drug to the list of drugs for that use
    for usage in drug_usages:
        alternatives_lookup[usage].append(drug_info)

end_time = time.time()
print(f"--- Processing complete in {end_time - start_time:.2f} seconds ---")
print(f"Found {len(alternatives_lookup)} unique usages.")
```

```python
try:
    with open('alternative_medicine.json', 'w') as f:
        json.dump(alternatives_lookup, f, indent=2)
    print("--- SUCCESS! ---")
    print(f"Your file is ready to be used by your FastAPI app.")
except Exception as e:
    print(f"ERROR: Could not save JSON file. {e}")
```

```python
print("Generating unique lists for autocomplete...")

unique_indications = sorted(list(alternatives_lookup.keys()))

unique_drugs = set()
for drug_list in alternatives_lookup.values():
    for drug in drug_list:
        unique_drugs.add(drug['name'])
unique_drugs = sorted(list(unique_drugs))

print(f"Found {len(unique_indications)} unique indications.")
print(f"Found {len(unique_drugs)} unique drugs.")

with open('alternatives_indications.json', 'w') as f:
    json.dump(unique_indications, f)
print("Saved new_indications.json")

with open('alternatives_drugs.json', 'w') as f:
    json.dump(unique_drugs, f)
print("Saved new_drugs.json")
```
