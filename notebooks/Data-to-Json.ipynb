```python
import pandas as pd
import numpy as np
```

```python
final_dataset = pd.read_csv('final_faers_data.csv')
final_dataset
```

```python
import json

unique_chemicals = final_dataset['active_chemical'].unique().tolist()
unique_chemicals.sort()
with open('active_chemicals.json', 'w') as f:
    json.dump(unique_chemicals, f)
print(f"Successfully saved {len(unique_chemicals)} chemicals to 'active_chemicals.json'")


unique_indications = final_dataset['indication_for_use'].unique().tolist()
unique_indications.sort()
with open('indications.json', 'w') as f:
    json.dump(unique_indications, f)
print(f"Successfully saved {len(unique_indications)} indications to 'indications.json'")


unique_routes = final_dataset['route'].unique().tolist()
unique_routes.sort()
with open('routes.json', 'w') as f:
    json.dump(unique_routes, f) 
print(f"Successfully saved {len(unique_routes)} routes to 'routes.json'")


reporter_countries = set(final_dataset['reporter_country'].unique())
occr_countries = set(final_dataset['occr_country'].unique())
all_unique_countries = list(reporter_countries.union(occr_countries))
all_unique_countries.sort() 
with open('countries.json', 'w') as f:
    json.dump(all_unique_countries, f)
print(f"Successfully saved {len(all_unique_countries)} unique countries to 'countries.json'")

print("\nDone! You can now move these JSON files to your 'static/js/' folder.")
```
