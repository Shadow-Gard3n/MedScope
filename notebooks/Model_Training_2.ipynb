```python
selected_cols = [
    'primaryid',
    
    'age_grp', 
    'sex', 
    'is_hcp',
    
    'active_chemical',
    'drug_role',
    'route',
    'indication_for_use',
    'dechallenge',
    
    # Target
    'outcome_code',
    'reaction_term'
]

df_basic = final_df[selected_cols].copy()
df_basic.head(10)
```

```python
df_basic.isnull().sum()
```

```python
df_basic['indication_for_use'] = df_basic['indication_for_use'].replace(
    'Product used for unknown indication', 'Unknown'
)
df_basic['indication_for_use'].value_counts()
```

```python
df_basic['drug_profile'] = df_basic['active_chemical'] + \
                           '_ROLE_' + df_basic['drug_role'] + \
                           '_ROUTE_' + df_basic['route'] + \
                           '_IND_' + df_basic['indication_for_use'] + \
                           '_DECHAL_' + df_basic['dechallenge']

df_basic
```

```python
df_agg = df_basic.groupby('primaryid').agg(
    age_grp=('age_grp', 'first'),
    sex=('sex', 'first'),
    is_hcp=('is_hcp', 'first'),
    
    all_drug_profiles=('drug_profile', lambda x: list(x.unique())),
    
    all_outcomes=('outcome_code', lambda x: list(x.unique())),
    all_reactions=('reaction_term', lambda x: list(x.unique()))
)

print(f"Original rows: {len(df_basic)}, Aggregated rows: {len(df_agg)}")
df_agg.head(10)
```

```python
# ['A_...', 'B_...'] -> "A_... B_..."
df_agg['drug_profile_joined'] = df_agg['all_drug_profiles'].apply(lambda x: ' '.join(x))

feature_columns = [
    'age_grp', 
    'sex', 
    'is_hcp',
    'drug_profile_joined'
]

X = df_agg[feature_columns]
X
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

# translate your list of text outcomes into a numeric matrix (filled with 0s and 1s)

outcome_classes = [
    'Other Serious', 'Hospitalization', 'Outcome Not Specified', 
    'Disability', 'Life-Threatening', 'Death', 
    'Congenital Anomaly', 'Required Intervention'
]

mlb_outcomes = MultiLabelBinarizer(classes=outcome_classes)

# Now for all the outcomes list in each row will be represented as list of 0 and 1 where 1 means this outcome is in the outcomes list
y_risk = mlb_outcomes.fit_transform(df_agg['all_outcomes']) 

y_risk
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.compose import ColumnTransformer

feature_preprocessor = ColumnTransformer(
    transformers=[
        # Handle 'low' cardinality categorical features
        ('cat_low', 
         OneHotEncoder(handle_unknown='ignore'), 
         ['age_grp', 'sex', 'is_hcp']),
        
        # Handle our complex drug profiles text
        ('drug_profiles_text', 
         TfidfVectorizer(max_features=5000), # Limit to top 5000 drug profiles
         'drug_profile_joined')
    ],
    remainder='drop' # Drop any columns we didn't specify
)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y_risk, test_size=0.2, random_state=42)

print(f"\nTraining on {len(X_train)} samples, testing on {len(X_test)} samples.")
```

```python
from sklearn.multioutput import MultiOutputClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline



base_classifier = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)

model_1 = MultiOutputClassifier(base_classifier)

pipeline_risk = Pipeline(steps=[
    ('preprocess', feature_preprocessor),
    ('model', model_1)
])

print("\n--- Starting Model 1 training... ---")
pipeline_risk.fit(X_train, y_train)
print("--- Model 1 Training Complete! ---")

print("\n--- Model 1 Evaluation ---")
y_pred = pipeline_risk.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print(f"Subset Accuracy (Strict): {accuracy_score(y_test, y_pred):.4f}")
print(f"F1 Score (Micro): {f1_score(y_test, y_pred, average='micro'):.4f}")
print(f"Recall (Micro): {recall_score(y_test, y_pred, average='micro'):.4f}")
print(f"Precision (Micro): {precision_score(y_test, y_pred, average='micro'):.4f}")

```

```python
custom_data = {
    'age_grp': 'Adult',
    'sex': 'M',
    'is_hcp': True,            
    'drug_profile_joined': 'ASPIRIN_ROLE_PS_ROUTE_Oral_IND_Headache_DECHAL_Yes'
}

custom_sample = pd.DataFrame(custom_data, index=[0])


prediction_encoded = pipeline_risk.predict(custom_sample)

prediction_labels = mlb_outcomes.inverse_transform(prediction_encoded)   # number to label

print("\n--- Custom Prediction Result ---")
print(f"Predicted Risk Profile: {prediction_labels[0]}")
```

## For reaction model

```python
top_200_reactions = df_basic['reaction_term'].value_counts().head(200).index.tolist()

mlb_reactions = MultiLabelBinarizer(classes=top_200_reactions)

y_reactions = mlb_reactions.fit_transform(df_agg['all_reactions'])

y_reactions
```

```python
X_train, X_test, y_train_reactions, y_test_reactions = train_test_split(
    X, y_reactions, test_size=0.2, random_state=42
)

print(f"\nTraining on {len(X_train)} samples, testing on {len(X_test)} samples.")
```

```python
base_classifier_2 = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)

model_2 = MultiOutputClassifier(base_classifier_2)

pipeline_reactions = Pipeline(steps=[
    ('preprocess', feature_preprocessor),
    ('model', model_2)
])

print("\n--- Starting Model 2 (Side Effect Profile) training... ---")
pipeline_reactions.fit(X_train, y_train_reactions)
print("--- Model 2 Training Complete! ---")
```

```python
print("\n--- Model 2 Evaluation ---")
y_pred_reactions = pipeline_reactions.predict(X_test)

print(f"Subset Accuracy (Strict): {accuracy_score(y_test_reactions, y_pred_reactions):.4f}")
print(f"F1 Score (Micro): {f1_score(y_test_reactions, y_pred_reactions, average='micro'):.4f}")
print(f"Recall (Micro): {recall_score(y_test_reactions, y_pred_reactions, average='micro'):.4f}")
print(f"Precision (Micro): {precision_score(y_test_reactions, y_pred_reactions, average='micro'):.4f}")
```
